```python
from pyspark.sql import SparkSession

catalog_name = "testUC" 

spark = (
    SparkSession.builder
    .appName("local-uc-test")
    .master("spark://spark:7077")
    .config("spark.jars.packages", "org.apache.hadoop:hadoop-aws:3.4.1,io.delta:delta-spark_2.12:3.3.1,io.unitycatalog:unitycatalog-spark_2.12:0.2.1")
    .config("spark.hadoop.fs.s3.impl", "org.apache.hadoop.fs.s3a.S3AFileSystem")
    .config("spark.sql.extensions", "io.delta.sql.DeltaSparkSessionExtension")
    .config("spark.sql.catalog.spark_catalog", "io.unitycatalog.spark.UCSingleCatalog")
    .config(f"spark.sql.catalog.{catalog_name}", "io.unitycatalog.spark.UCSingleCatalog")
    .config(f"spark.sql.catalog.{catalog_name}.uri", "http://server:8080")
    .config(f"spark.sql.catalog.{catalog_name}.token", "")
    .config("spark.sql.defaultCatalog", catalog_name)
    .getOrCreate()
)
```

```python
# Create new schema
# spark.sql("CREATE SCHEMA demo")

# Should now show two schemas: default and demo
spark.sql("SHOW SCHEMAS").show()
```

```python
# # Create a new table
new_tablename = f"testUC.demo.sctable2"
# spark.sql(f"DROP TABLE {new_tablename}")
spark.sql("USE SCHEMA demo")

spark.sql("SHOW SCHEMAS").show()
spark.sql("SELECT current_catalog(), current_schema()").show()
spark.sql("SHOW SCHEMAS FROM testUC").show()

print(new_tablename)

spark.sql(f"""
CREATE TABLE `testUC`.`demo`.`sctable2` (id INT, desc STRING)
USING delta
""")
# Insert new rows
spark.sql(f"INSERT INTO {new_tablename} VALUES (1, 'test 1')")
spark.sql(f"INSERT INTO {new_tablename} VALUES (2, 'test 2')")
spark.sql(f"INSERT INTO {new_tablename} VALUES (3, 'test 3')")
spark.sql(f"INSERT INTO {new_tablename} VALUES (4, 'test 4')")

# Read table
spark.sql(f"SELECT * FROM {new_tablename}").show()
```

```python
# # Create a new table
spark.sql("""
CREATE TABLE testShema.sparkTable1 (id INT, desc STRING)
USING delta
""")
```

```python
csv_path = "/opt/data/example.csv"
delta_path = "/opt/data/deltaTable1"
df = spark.read.option("header", "true").option("inferSchema", "true") .csv(csv_path)

df.write.format("delta").mode("overwrite").saveAsTable("testUC.testShema.autoTable1", path=delta_path)
```

```python
spark.sql("SELECT * FROM testShema.autoTable1").show()
```
